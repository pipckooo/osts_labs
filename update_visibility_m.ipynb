```python
from sqlalchemy.orm import Session
from sqlalchemy import create_engine, update, Table, MetaData

engine = create_engine('postgresql+psycopg2://postgres:pass@localhost/traffic_db')

metadata = MetaData()
traffic_table = Table("traffic_accident_records_updated", metadata, autoload_with=engine)

with Session(engine) as session:
    
    session.execute(update(traffic_table).where(traffic_table.c.visibility_m==-11).values(visibility_m=-1))
    
    session.commit()

print("session has been committed")
```
