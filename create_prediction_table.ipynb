```python
from sqlalchemy import Column, Integer, MetaData, String, Table, create_engine


engine = create_engine('postgresql+psycopg2://postgres:pass@localhost/traffic_db')

metadata = MetaData()


predictions = Table(
    'predictions',
    metadata,
    Column('prediction_id', Integer, primary_key=True, autoincrement=True),
    Column('prediction_value', String, nullable=False),
    Column('row_id', String, nullable=False),
    Column('real_value', String, nullable=False),
    Column('source_type', String, nullable=False),
    Column('model_type', String, nullable=False),
)


metadata.create_all(engine)
print("Table created .")
```
