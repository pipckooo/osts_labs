```python
from sqlalchemy.orm import Session
from sqlalchemy import create_engine, update, Table, MetaData

engine = create_engine('postgresql+psycopg2://postgres:pass@localhost/traffic_db')

metadata = MetaData()
traffic_table = Table("traffic_accident_records_updated", metadata, autoload_with=engine)

with Session(engine) as session:

    session.execute(update(traffic_table).where(traffic_table.c.day_of_week == 'Monday').values(monday=1))
    session.execute(update(traffic_table).where(traffic_table.c.day_of_week == 'Tuesday').values(tuesday=1))
    session.execute(update(traffic_table).where(traffic_table.c.day_of_week == 'Wednesday').values(wednesday=1))
    session.execute(update(traffic_table).where(traffic_table.c.day_of_week == 'Thursday').values(thursday=1))
    session.execute(update(traffic_table).where(traffic_table.c.day_of_week == 'Friday').values(friday=1))  
    session.commit()

print("session has been committed")
```
