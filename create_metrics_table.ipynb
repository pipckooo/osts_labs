```python
from sqlalchemy import Column, Integer, Float, MetaData, String, Table, create_engine, DateTime, func

engine = create_engine('postgresql+psycopg2://postgres:pass@localhost/traffic_db')

metadata = MetaData()

model_metrics = Table(
    'model_metrics',
    metadata,
    Column('metric_id', Integer, primary_key=True, autoincrement=True),
    Column('model_type', String, nullable=False),
    Column('source_type', String, nullable=False),   
    Column('recall', Float, nullable=False),
    Column('log_loss', Float, nullable=False),
    Column('precision', Float, nullable=False),
    Column('f1_score', Float, nullable=False),
    Column('created_at', DateTime, server_default=func.now()),
)

metadata.create_all(engine)
print("Table created.")
```
