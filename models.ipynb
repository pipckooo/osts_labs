```python
from sqlalchemy import Column, String, DateTime, Date, Integer, Boolean, Numeric, text
from sqlalchemy.orm import declarative_base

Base = declarative_base()

class TrafficIncident(Base):

    __tablename__ = 'traffic_incidents'

 
    record_id = Column(String(64), primary_key=True) 
    timestamp = Column(DateTime) 
    record_date = Column(Date) 
    hour = Column(Integer) 
    month = Column(Integer) 
    day_of_week = Column(String(15)) 
    time_of_day = Column(String(20)) 
    is_holiday = Column(Boolean) 
    area_type = Column(String(50))
    road_type = Column(String(50)) 
    intersection_type = Column(String(50)) 
    speed_limit_kmh = Column(Integer) 
    weather = Column(String(50)) 
    road_condition = Column(String(50)) 
    
    
    visibility_m = Column(Numeric(8, 2))
    congestion_index = Column(Numeric(5, 2))
    vehicle_type = Column(String(50))
    driver_age = Column(Numeric(4, 1))
    driver_experience_yrs = Column(Numeric(4, 1))
    vehicle_age_yrs = Column(Numeric(4, 1))
    recorded_speed_kmh = Column(Numeric(5, 2))
    baseline_expected_response_time_min = Column(Numeric(6, 2))
    
    accident_occurred = Column(Boolean) 
    collision_type = Column(String(50)) 
    vehicles_involved = Column(Integer) 
    accident_severity = Column(String(50)) 
    dispatch_priority = Column(String(20)) 
    actual_response_time_min = Column(Numeric(6, 2)) 
    road_closure_duration_min = Column(Numeric(7, 2)) 


    urban_area = Column(Boolean, server_default=text('false'))
    suburban_area = Column(Boolean, server_default=text('false'))
    rural_area = Column(Boolean, server_default=text('false'))
    arterial_road_type = Column(Boolean, server_default=text('false'))
    local_road_type = Column(Boolean, server_default=text('false'))
    interstate_road_type = Column(Boolean, server_default=text('false'))
    collector_road_type = Column(Boolean, server_default=text('false'))
    four_way_intersection_type = Column(Boolean, server_default=text('false'))
    t_junction_intersection_type = Column(Boolean, server_default=text('false'))
    roundabout_intersection_type = Column(Boolean, server_default=text('false'))
    clear_weather = Column(Boolean, server_default=text('false'))
    rain = Column(Boolean, server_default=text('false'))
    fog = Column(Boolean, server_default=text('false'))
    snow = Column(Boolean, server_default=text('false'))
    dry_road = Column(Boolean, server_default=text('false'))
    wet_road = Column(Boolean, server_default=text('false'))
    snowy_road = Column(Boolean, server_default=text('false'))
    icy_road = Column(Boolean, server_default=text('false'))
    standing_water = Column(Boolean, server_default=text('false'))
    sedan = Column(Boolean, server_default=text('false'))
    bus = Column(Boolean, server_default=text('false'))
    motorcycle = Column(Boolean, server_default=text('false'))
    truck = Column(Boolean, server_default=text('false'))
    suv = Column(Boolean, server_default=text('false'))
    head_on_collision = Column(Boolean, server_default=text('false'))
    single_vehicle_collision = Column(Boolean, server_default=text('false'))
    side_impact_collision = Column(Boolean, server_default=text('false'))
    rear_end_collision = Column(Boolean, server_default=text('false'))
    minor_accident = Column(Boolean, server_default=text('false'))
    severe_accident = Column(Boolean, server_default=text('false'))
    moderate_accident = Column(Boolean, server_default=text('false'))
    fatal_accident = Column(Boolean, server_default=text('false'))
    standard_dispatch = Column(Boolean, server_default=text('false')) 
    high_dispatch = Column(Boolean, server_default=text('false'))
    medium_dispatch = Column(Boolean, server_default=text('false'))
```
